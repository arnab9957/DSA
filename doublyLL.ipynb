```python
class Node:
    def __init__(self, data=None):
        self.data = data
        self.next = None
        self.prev = None

class DoublyLL:
    def __init__(self, head=None):
        self.head = head

    def insert_at_end(self, value):
        new_node = Node(value)

        if(self.head != None):
            temp = self.head
            while(temp.next != None):
                temp = temp.next
            temp.next = new_node
            new_node.prev = temp
        else:
            self.head = new_node
            return

    def insert_at_begining(self, value):
        new_node = Node(value)
        new_node.next = self.head
        self.head = new_node
        self.head.prev = new_node

    def insert_at_any(self, value, find):
        new_node = Node(value)
        temp = self.head
        while temp != None:
            if (temp.data == find):
                new_node.next = temp.next
                new_node.prev = temp
                temp.next = new_node
                break
            temp = temp.next

    def deleteNode(self, value):
            temp = self.head
            prev = temp

            if(temp.data == value):
                self.head = temp.next
                self.head.prev = None

            while temp != None:
                if(temp.data == value):
                    prev.next = temp.next
                    temp.prev = prev
                    break
                else:
                    prev = temp
                    temp = temp.next

    def PrintLL(self):
        temp = self.head
        while(temp != None):
            print(temp.data, end=" <--> ")
            temp = temp.next



L = DoublyLL()
L.insert_at_end(10)
L.insert_at_end(20)
L.insert_at_end(30)
L.insert_at_end(40)
L.insert_at_begining(5)
L.insert_at_begining(4)
L.insert_at_any(35, 30)
L.insert_at_any(25, 20)
L.deleteNode(40)
L.deleteNode(30)
L.deleteNode(4)
L.PrintLL()
```

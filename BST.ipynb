```python
#Assigning values for pointers of tree nodes (BINARY SEARCH TREEE)
class Node:
    def __init__(self, values):
        self.data = values
        self.left = None
        self.right = None

def search(root, value):
    if (root == None):
        print("Not Found?!😒")
        return
    if (root.data == value):
        print("Found!!🥱")
        return
    if (root.data < value):
        search(root.right, value) 
    else:
        search(root.left, value)

def insert(root, value):
    if (root == None):
        return Node(value)
    if (root.data == value):
        return root
    if (root.data < value):
        root.right = insert(root.right, value) 
    else:
        root.left = insert(root.left, value)
    return root

def get_successor(root):
    root = root.right
    while (root == None and root.left == None):
        root = root.left
    return root

def delete(root, value):
    if (root == None):
        return root
    if(root.data > value):
        root.left = delete(root.left, value)
    if(root.data < value):
        root.right = delete(root.right, value)

    else:
        if (root.left == None):
            return root.right
        if (root.right == None):
            return root.left
        else:
            succ = get_successor(root)
            root.data = succ.data 
            root.left = delete(root.right, succ.data)


#Inserting values in the tree
# root = Node(10)
# root.left = Node(5)
# root.right = Node(15)
# root.left.left = Node(3)
# root.left.right = Node(7)

root = insert(None, 10) #to prevent modification inn root we are asigning again
root = insert(root, 5)
root = insert(root, 15)
root = insert(root, 3)
root = insert(root, 60)
root = insert(root, 49)

print("\n")


search(root, 5)
search(root, 55)



# def preorder_traversal(root):
#     if (root != None):
#         print(f"preorder_traversal: {root.data}")
#         preorder_traversal(root.left)
#         preorder_traversal(root.right)
def inorder_traversal(root):
    if (root != None):
        inorder_traversal(root.left)
        print(f"Inorder_traversal: {root.data}")
        inorder_traversal(root.right)

inorder_traversal(root)
print("\n")
delete(root, 15)
inorder_traversal(root)

# preorder_traversal(root)
```

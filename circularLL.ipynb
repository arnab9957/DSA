```python
class Node:
    def __init__(self, data, next=None):
        self.data = data
        self.next = next


class CircularSinglyLL:
    def __init__(self, head=None):
        self.head = head

    def insert_at_end(self, value):
        new_node = Node(value)

        if self.head is None:
            self.head = new_node
            new_node.next = self.head
        else:
            temp = self.head
            while temp.next != self.head:
                temp = temp.next
            temp.next = new_node
            new_node.next = self.head

    def insert_at_begining(self, value):
        new_node = Node(value)

        if self.head is None:
            self.head = new_node
            new_node.next = self.head
        else:
            temp = self.head
            while temp.next != self.head:
                temp = temp.next
            new_node.next = self.head
            temp.next = new_node
            self.head = new_node

    def insert_at_any(self, value, find):
        if self.head is None:
            return

        new_node = Node(value)
        temp = self.head

        while True:
            if temp.data == find:
                new_node.next = temp.next
                temp.next = new_node
                return
            temp = temp.next
            if temp == self.head:
                break

    def deleteNode(self, value):
        if self.head is None:
            return

        temp = self.head
        prev = None

        if self.head.data == value:
            if self.head.next == self.head:
                self.head = None
                return

            prev = self.head
            while prev.next != self.head:
                prev = prev.next
            self.head = self.head.next
            prev.next = self.head
            return

        prev = self.head
        temp = self.head.next

        while temp != self.head:
            if temp.data == value:
                prev.next = temp.next
                return
            prev = temp
            temp = temp.next

    def PrintLL(self):
        if self.head is None:
            print("List is empty")
            return

        temp = self.head
        while True:
            print(temp.data, end=" ")
            temp = temp.next
            if temp == self.head:
                break
        print()


L = CircularSinglyLL()
L.insert_at_end(10)
L.insert_at_end(20)
L.insert_at_end(30)
L.insert_at_end(40)
L.insert_at_begining(5)
L.insert_at_any(25, 20)
L.deleteNode(40)
L.deleteNode(30)
L.PrintLL()
```

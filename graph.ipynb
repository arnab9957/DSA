```python
#Adjacency List

class Graph:
    def __init__(self):
        self.adjlist = {}

    def add_vertex(self, vertex):
        if vertex not in self.adjlist:
            self.adjlist[vertex] = []

    def add_edge(self, src, dest):
        self.add_vertex(src)
        self.add_vertex(dest)

        self.adjlist[src].append(dest)
        self.adjlist[dest].append(src)

    def Print_Graph(self):
        for vertex in self.adjlist:
            print(vertex, " --> ", self.adjlist[vertex], end="\n")




G = Graph()

G.add_vertex(5)
G.add_edge(1,2)
G.add_edge(2,3)
G.add_edge(4,3)
G.add_edge(5,2)
G.add_edge(1,5)
G.add_edge(1,4)
G.Print_Graph()

```

```python
#DFS

class Graph:
    def __init__(self, vertex):
        self.matrix = [ [0]*vertex for _ in range(vertex) ]
        self.size = vertex

    def add_edge(self, start, end):
        if (0 <= start, end <= self.size):
            self.matrix[start][end] = 1
            self.matrix[end][start] = 1
        else:
            print("Invalid Edge...😒")

    def print(self):
        for row in self.matrix:
            print(' '.join(map(str, row)))

    def dfs(self, src):
        visited = [False] * self.size
        stack = [src] #insert the first element in stack

        while(stack):
            poped = stack.pop()
            
            if(visited[poped] == False):
                print(poped, end=" --> ")
                visited[poped] = True

            #now i hvae to check that from that poped node how many nodes there is:
            for vertex in range(self.size):
                if(self.matrix[poped][vertex] == 1 and visited[vertex] == False):
                    stack.append(vertex)

G = Graph(6)

G.add_edge(0,1)
G.add_edge(0,2)
G.add_edge(2,3)
G.add_edge(2,4)
G.add_edge(3,5)
G.add_edge(4,5)

G.dfs(0)

```

```python
#BFS
from collections import deque

class Graph:
    def __init__(self, vertex):
        self.matrix = [ [0]*vertex for _ in range(vertex) ]
        self.size = vertex

    def add_edge(self, start, end):
        if (0 <= start, end <= self.size):
            self.matrix[start][end] = 1
            self.matrix[end][start] = 1
        else:
            print("Invalid Edge...😒")

    def print(self):
        for row in self.matrix:
            print(' '.join(map(str, row)))

    def BFS(self, src):
        visited = [False] * self.size
        queue = deque([src]) #insert the first element in queue
        visited[src]= True
        while(queue):
            poped = queue.popleft()
            print(poped, end=" --> ")

            #now i hvae to check that from that poped node how many nodes there is:
            for vertex in range(self.size):
                if(self.matrix[src][vertex] == 1 and visited[vertex] == False):
                    queue.append(vertex)
                    visited[vertex] = True

G = Graph(6)

G.add_edge(0,1)
G.add_edge(0,2)
G.add_edge(2,3)
G.add_edge(2,4)
G.add_edge(3,5)
G.add_edge(4,5)

G.BFS(0)
```
